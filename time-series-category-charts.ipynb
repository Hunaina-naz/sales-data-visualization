```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dff = pd.read_csv('/kaggle/input/sample-dataset/SampleSuperstore.csv')
dff
```

```python
print(dff.columns)  # i just checked columns just to easy for me
```

```python
dff.head()  # first five rows from data frame
```

# Create Synthetic Date Column¶

```python
print(dff['Date'].head())
```

```python
dff['Date'] = pd.date_range(start='2023-01-01', periods=len(dff), freq='D')
```

```python
print(dff.columns)
```

```python
# Remove any hidden spaces from column names
dff.rename(columns=lambda x: x.strip(), inplace=True)
print('Date' in dff.columns)  # Should print True
```

```python
dff.set_index('Date', inplace=True)
print(dff.head())
```

# Aggregate Sales Monthly

```python
monthly_sales = dff['Sales'].resample('M').sum()   #“Sales were aggregated monthly using pandas resample with Month-End frequency to analyze trends over time.”
print(monthly_sales.head())
```

# Line Chart – Monthly Sales Trend

```python
plt.figure()                 # Line chart showing monthly sales trend over time.
monthly_sales.plot()         # X-axis: Months, Y-axis: Total Sales.
plt.title("Monthly Sales Trend")   # This visualization highlights trends and patterns in sales over the months.
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True)
plt.tight_layout()
plt.savefig("monthly_sales_trend.png")
plt.show()  
```

# Bar Chart – Sales by Category

```python
category_sales = dff.groupby('Category')['Sales'].sum()    # Category sales

plt.figure()
category_sales.plot(kind='bar')
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("category_sales_bar.png")
plt.show()
```

## Pie Chart – Sales Share by Category

```python
plt.figure()       #Helps identify dominant category
category_sales.plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title("Sales Share by Category")
plt.ylabel("")
plt.tight_layout()
plt.savefig("category_sales_pie.png")
plt.show()
```

# A Short description
Monthly sales trends were visualized using a line chart after aggregating sales data by month. Bar charts were used to compare total sales across product categories, revealing relative performance. A pie chart illustrated the percentage share of each category in total sales. All charts include proper labels and formatting for clarity.
